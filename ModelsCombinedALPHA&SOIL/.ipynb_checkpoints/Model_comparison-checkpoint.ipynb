Imports


```python
import pandas as pd
import numpy as np
import joblib
import matplotlib.pyplot as plt

from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.linear_model import Ridge
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.model_selection import (
    GroupKFold, LeaveOneGroupOut, learning_curve, validation_curve,
)
from sklearn.metrics import r2_score

RANDOM_STATE = 42
MAX_LODO_FOLDS = 15
CROPS = ["Barley", "Buckwheat", "Maize", "Millet", "Wheat"]
CURVE_CROPS = ["Maize", "Wheat"]  # only crops with real signal get learning/validation curves
```

## 2. Load data & build features

```python
df = pd.read_csv( r"C:\Users\SRDC\Desktop\Alpha_earth\merged_dataset\nepal_merged_dataset.csv" )
df.columns = [c.strip() for c in df.columns]
for c in df.columns:
    if df[c].dtype == object:
        df[c] = df[c].str.strip()
for crop in CROPS:
    for metric in ["Area_Ha", "Production_Mt", "Yield_MtHa"]:
        col = f"{crop}_{metric}"
        df[col] = pd.to_numeric(df[col], errors="coerce")

embed_cols = [c for c in df.columns if c.startswith("A") and c[1:].isdigit()]
numeric_feat_cols = embed_cols + [
    "Area_sqkm", "Elevation_Min", "Elevation_Max", "Slope_Median", "Relief_Median",
    "Potassium_mean", "Nitrogen_mean", "OrganicMatter_mean", "Phosphorous_mean", "pH_mean",
]
categorical_feat_cols = ["Landform", "Parent_Material", "Dominant_Soil"]
X_all = pd.get_dummies(df[numeric_feat_cols + categorical_feat_cols], columns=categorical_feat_cols)
print("Feature matrix shape:", X_all.shape)
```

```python
prior = pd.read_csv("crop_yield_full_validation_results.csv")
lodo = prior[prior["CV_Scheme"] == "Leave-One-District-Out"]
best_model_per_crop = lodo.loc[lodo.groupby("Crop")["R2"].idxmax()].set_index("Crop")["Model"].to_dict()
print("Best model per crop (by LODO R2):", best_model_per_crop)


def make_model(name):
    if name == "Ridge":
        return Pipeline([("scaler", StandardScaler()), ("model", Ridge(alpha=1.0, random_state=RANDOM_STATE))])
    if name == "RandomForest":
        return RandomForestRegressor(n_estimators=150, max_depth=8, random_state=RANDOM_STATE, n_jobs=-1)
    if name == "XGBoost":
        return XGBRegressor(n_estimators=150, max_depth=4, learning_rate=0.08,
                             random_state=RANDOM_STATE, n_jobs=-1, verbosity=0)
    raise ValueError(name)


def get_crop_data(crop):
    target_col = f"{crop}_Yield_MtHa"
    mask = df[target_col].notna()
    y = df.loc[mask, target_col].reset_index(drop=True)
    X = X_all.loc[mask].reset_index(drop=True)
    groups = df.loc[mask, "District_Key"].reset_index(drop=True)
    return X, y, groups
```

## 3. Pick best model per crop (by prior LODO R2)

## 4. Train final deployable models
Fit on **100% of available data** per crop -- these are for making new predictions, not for reporting generalization performance (use the CV results from the earlier notebooks for that).

```python
final_models = {}
for crop in CROPS:
    X, y, _ = get_crop_data(crop)
    for model_name in ["RandomForest", "XGBoost"]:
        model = make_model(model_name)
        model.fit(X, y)
        final_models[(crop, model_name)] = model
        joblib.dump(model, f"final_model_{crop}_{model_name}.joblib")
    print(f"{crop}: trained on {len(y)} samples, saved RandomForest + XGBoost models")
```

## 5. Feature importance (best model per crop)

```python
fig, axes = plt.subplots(1, len(CROPS), figsize=(5 * len(CROPS), 5))
for ax, crop in zip(axes, CROPS):
    best_name = best_model_per_crop[crop]
    if best_name == "Ridge":
        best_name = "RandomForest"  # Ridge has coefficients not importances; use RF for viz consistency
    model = final_models[(crop, best_name)]
    importances = pd.Series(model.feature_importances_, index=X_all.columns).sort_values(ascending=False).head(15)
    importances.iloc[::-1].plot(kind="barh", ax=ax)
    ax.set_title(f"{crop}\n(best: {best_model_per_crop[crop]}, shown: {best_name})")
    ax.set_xlabel("Importance")
plt.tight_layout()
plt.savefig("feature_importance_all_crops.png", dpi=150)
plt.show()
```

## 6. Predicted vs Actual
Using **out-of-fold** CV predictions (LODO scheme) -- this is the honest generalization view, not an in-sample fit.

```python
fig, axes = plt.subplots(1, len(CROPS), figsize=(5 * len(CROPS), 5))
for ax, crop in zip(axes, CROPS):
    X, y, groups = get_crop_data(crop)
    best_name = best_model_per_crop[crop]
    n_groups = groups.nunique()
    splitter = LeaveOneGroupOut() if n_groups <= MAX_LODO_FOLDS else GroupKFold(n_splits=MAX_LODO_FOLDS)
    y_true_all, y_pred_all = [], []
    for train_idx, test_idx in splitter.split(X, y, groups):
        m = make_model(best_name)
        m.fit(X.iloc[train_idx], y.iloc[train_idx])
        preds = m.predict(X.iloc[test_idx])
        y_true_all.extend(y.iloc[test_idx].tolist())
        y_pred_all.extend(preds.tolist())
    y_true_all, y_pred_all = np.array(y_true_all), np.array(y_pred_all)
    r2 = r2_score(y_true_all, y_pred_all)

    ax.scatter(y_true_all, y_pred_all, alpha=0.5, s=20)
    lims = [min(y_true_all.min(), y_pred_all.min()), max(y_true_all.max(), y_pred_all.max())]
    ax.plot(lims, lims, "k--", linewidth=1, label="Perfect prediction")
    ax.set_xlabel("Actual Yield (Mt/Ha)")
    ax.set_ylabel("Predicted Yield (Mt/Ha)")
    ax.set_title(f"{crop} ({best_name})\nR2={r2:.3f}")
    ax.legend(fontsize=8)
plt.tight_layout()
plt.savefig("predicted_vs_actual_all_crops.png", dpi=150)
plt.show()
```

## 7. Learning curves (Maize, Wheat)
Score vs. training set size, grouped 5-fold CV by district. If the curve is still rising at max sample size, the model is data-starved, not maxed out.

```python
fig, axes = plt.subplots(1, len(CURVE_CROPS), figsize=(6 * len(CURVE_CROPS), 5))
for ax, crop in zip(axes, CURVE_CROPS):
    X, y, groups = get_crop_data(crop)
    for model_name, color in [("RandomForest", "tab:blue"), ("XGBoost", "tab:orange")]:
        model = make_model(model_name)
        train_sizes, train_scores, test_scores = learning_curve(
            model, X, y, groups=groups, cv=GroupKFold(n_splits=5),
            train_sizes=np.linspace(0.2, 1.0, 5), scoring="r2", n_jobs=-1,
        )
        ax.plot(train_sizes, test_scores.mean(axis=1), "o-", color=color, label=f"{model_name} (val)")
        ax.fill_between(train_sizes, test_scores.mean(axis=1) - test_scores.std(axis=1),
                         test_scores.mean(axis=1) + test_scores.std(axis=1), alpha=0.15, color=color)
    ax.set_xlabel("Training set size (samples)")
    ax.set_ylabel("R2 (grouped 5-fold CV)")
    ax.set_title(f"{crop}: Learning Curve")
    ax.axhline(0, color="black", linewidth=0.5)
    ax.legend(fontsize=8)
plt.tight_layout()
plt.savefig("learning_curves_maize_wheat.png", dpi=150)
plt.show()
```

## 8. Validation curves (max_depth, Maize, Wheat)
Score vs. `max_depth` -- shows whether the model is under/overfitting at the chosen hyperparameter value.

```python
param_range = [2, 4, 6, 8, 12]
fig, axes = plt.subplots(1, len(CURVE_CROPS), figsize=(6 * len(CURVE_CROPS), 5))
for ax, crop in zip(axes, CURVE_CROPS):
    X, y, groups = get_crop_data(crop)
    for model_name, base_model, color in [
        ("RandomForest", RandomForestRegressor(n_estimators=100, random_state=RANDOM_STATE, n_jobs=-1), "tab:blue"),
        ("XGBoost", XGBRegressor(n_estimators=100, learning_rate=0.08, random_state=RANDOM_STATE, n_jobs=-1, verbosity=0), "tab:orange"),
    ]:
        train_scores, test_scores = validation_curve(
            base_model, X, y, param_name="max_depth", param_range=param_range,
            groups=groups, cv=GroupKFold(n_splits=5), scoring="r2", n_jobs=-1,
        )
        ax.plot(param_range, test_scores.mean(axis=1), "o-", color=color, label=f"{model_name} (val)")
        ax.fill_between(param_range, test_scores.mean(axis=1) - test_scores.std(axis=1),
                         test_scores.mean(axis=1) + test_scores.std(axis=1), alpha=0.15, color=color)
    ax.set_xlabel("max_depth")
    ax.set_ylabel("R2 (grouped 5-fold CV)")
    ax.set_title(f"{crop}: Validation Curve (max_depth)")
    ax.axhline(0, color="black", linewidth=0.5)
    ax.legend(fontsize=8)
plt.tight_layout()
plt.savefig("validation_curves_maize_wheat.png", dpi=150)
plt.show()
```
