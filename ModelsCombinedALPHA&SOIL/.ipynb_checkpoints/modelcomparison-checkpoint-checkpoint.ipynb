# Compare Ridge vs Random Forest vs XGBoost
Loads the three per-model results CSVs (produced by `ridge_model.ipynb`, `xgboost_model.ipynb`, `random_forest_model.ipynb`) and builds a single side-by-side comparison: full merged table, best-model-per-crop summary, and a grouped bar chart of R² by crop and model.

## 1. Imports & config

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.width", 160)
pd.set_option("display.max_columns", 20)

RESULT_FILES = {
    "Ridge": "ridge_results.csv",
    "RandomForest": "random_forest_results.csv",
    "XGBoost": "xgboost_results.csv",
}
CROPS = ["Barley", "Buckwheat", "Maize", "Millet", "Wheat"]
CV_SCHEME_TO_COMPARE = "Leave-One-District-Out"  # the harder, more realistic generalization test
```

## 2. Load and merge all three result sets

```python
frames = []
for model_name, path in RESULT_FILES.items():
    try:
        d = pd.read_csv(path)
        frames.append(d)
    except FileNotFoundError:
        print(f"WARNING: {path} not found -- run {model_name.lower()}_model.ipynb first to generate it.")

all_results = pd.concat(frames, ignore_index=True)
print(f"Loaded {len(all_results)} rows across {all_results['Model'].nunique()} models")
all_results.head()
```

## 3. Full comparison table (R² by crop × model × CV scheme)

```python
comparison = all_results.pivot_table(
    index=["Crop", "CV_Scheme"], columns="Model", values="R2"
).loc[pd.IndexSlice[CROPS, :], :]
comparison
```

## 4. Best model per crop
Ranked on the harder, more realistic **Leave-One-District-Out** score (a model that only looks good on Leave-One-Year-Out is likely overfitting to district-level patterns it memorized rather than learning something that transfers).

```python
lodo = all_results[all_results["CV_Scheme"] == CV_SCHEME_TO_COMPARE]
best_per_crop = (
    lodo.loc[lodo.groupby("Crop")["R2"].idxmax()]
    [["Crop", "Model", "R2", "RMSE", "MAE", "MAPE_pct"]]
    .set_index("Crop")
    .loc[CROPS]
)
print(f"Best model per crop, by {CV_SCHEME_TO_COMPARE} R2:\n")
best_per_crop
```

## 5. Grouped bar chart — R² by crop, all 3 models, both CV schemes

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=True)

for ax, scheme in zip(axes, ["Leave-One-Year-Out", "Leave-One-District-Out"]):
    subset = all_results[all_results["CV_Scheme"] == scheme]
    pivot = subset.pivot(index="Crop", columns="Model", values="R2").loc[CROPS]
    pivot.plot(kind="bar", ax=ax, legend=(scheme == "Leave-One-District-Out"))
    ax.axhline(0, color="black", linewidth=0.8)
    ax.set_title(scheme)
    ax.set_ylabel("R2" if scheme == "Leave-One-Year-Out" else "")

plt.suptitle("Out-of-fold R2 by crop, model, and CV scheme")
plt.tight_layout()
plt.show()
```

## 6. Where do models agree / disagree?
Crops where all three models land on the same side of zero are the trustworthy signal (Maize, Wheat). Crops where models disagree or all sit near/below zero (Barley, Millet, Buckwheat) are the ones limited by data sparsity, not by model choice -- switching models won't fix them, more/better data would.

```python
agreement = comparison.loc[pd.IndexSlice[:, CV_SCHEME_TO_COMPARE], :].droplevel("CV_Scheme")
agreement["All_positive"] = (agreement > 0).all(axis=1)
agreement["Spread"] = agreement[["Ridge", "RandomForest", "XGBoost"]].max(axis=1) - agreement[["Ridge", "RandomForest", "XGBoost"]].min(axis=1)
agreement.sort_values("All_positive", ascending=False)
```

## 7. Save combined comparison table

```python
comparison.to_csv("model_comparison.csv")
best_per_crop.to_csv("best_model_per_crop.csv")
print("Saved model_comparison.csv and best_model_per_crop.csv")
```
