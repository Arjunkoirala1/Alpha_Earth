# Spatially-Blocked LODO Cross-Validation

**Purpose:** Naive Leave-One-District-Out (LODO) treats every district as an independent fold. But two districts that are geographically or featurally near-identical (e.g. Mustang & Manang) usually end up on opposite sides of a train/test split — so the model isn't really extrapolating to new geography, it's interpolating between near-duplicates. This inflates LODO R².

**Fix:** cluster districts into spatial/feature **blocks** first, then run GroupKFold using the block ID as the group. This guarantees a held-out district's nearest neighbors are held out *with* it, so training can never contain a near-duplicate of a test point.

This notebook:
1. Clusters districts using lat/long (default) or AlphaEarth embedding centroid.
2. Runs GroupKFold-by-cluster instead of GroupKFold-by-district.
3. Checks, for every fold, that the minimum train-test distance exceeds a buffer radius — flags the fold if not.
4. Reports naive per-district LODO R² next to the blocked R², so the gap itself is a reportable number.


```
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.model_selection import GroupKFold
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score
from sklearn.preprocessing import StandardScaler
from scipy.spatial.distance import cdist
```

## 1. Config — edit these paths/columns for your dataset

```
DATASET_PATH   = "your_combined_dataset.csv"   # <-- change this
CROP           = "Wheat"
TARGET_COL     = "Wheat_yield"                  # <-- change per crop
DISTRICT_COL   = "District"
LAT_COL        = "Latitude"
LON_COL        = "Longitude"

CLUSTER_ON       = "latlong"   # "latlong" or "embedding"
EMBEDDING_PREFIX = "A"         # AlphaEarth cols are usually A00..A63
N_CLUSTERS       = 12
BUFFER_KM        = 15.0
MODEL_NAME       = "ridge"     # "ridge" or "rf"
RANDOM_STATE     = 42
```

## 2. Load data and pick feature columns

```
df = pd.read_csv(DATASET_PATH)

embedding_cols = [c for c in df.columns if c.startswith(EMBEDDING_PREFIX)]
non_feature_cols = {TARGET_COL, DISTRICT_COL, LAT_COL, LON_COL, "Year", "block"}
feature_cols = [c for c in df.columns
                if c not in non_feature_cols and pd.api.types.is_numeric_dtype(df[c])]

print(f"{len(feature_cols)} feature columns, {df[DISTRICT_COL].nunique()} unique districts")
df.head()
```

## 3. Clustering districts into spatial blocks

Clusters are computed on **unique districts**, not district-year rows, so a district's block assignment stays fixed across all its fiscal years.

```
def haversine_km(lat1, lon1, lat2, lon2):
    """Vectorized haversine distance in km."""
    r = 6371.0
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = np.sin(dlat / 2) ** 2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2) ** 2
    return 2 * r * np.arcsin(np.sqrt(a))


def build_blocks(df, district_col, lat_col, lon_col, n_clusters,
                  cluster_on="latlong", embedding_cols=None, random_state=42):
    district_level = df.drop_duplicates(subset=[district_col]).copy()

    if cluster_on == "latlong":
        X = district_level[[lat_col, lon_col]].values
        X = StandardScaler().fit_transform(X)
    elif cluster_on == "embedding":
        if not embedding_cols:
            raise ValueError("embedding_cols required when cluster_on='embedding'")
        X = district_level.groupby(district_col)[embedding_cols].mean().values
        X = StandardScaler().fit_transform(X)
    else:
        raise ValueError("cluster_on must be 'latlong' or 'embedding'")

    km = KMeans(n_clusters=n_clusters, random_state=random_state, n_init=10)
    district_level["block"] = km.fit_predict(X)

    block_map = dict(zip(district_level[district_col], district_level["block"]))
    df = df.copy()
    df["block"] = df[district_col].map(block_map)
    return df, block_map
```

## 4. Leakage check — verify held-out points are actually far from train

```
def min_train_test_distance_km(df, train_idx, test_idx, lat_col, lon_col):
    train_coords = df.loc[train_idx, [lat_col, lon_col]].drop_duplicates().values
    test_coords = df.loc[test_idx, [lat_col, lon_col]].drop_duplicates().values
    dists = cdist(test_coords, train_coords,
                   lambda u, v: haversine_km(u[0], u[1], v[0], v[1]))
    return dists.min()
```

## 5. Blocked cross-validation (leave-one-block-out)

```
def run_blocked_cv(df, feature_cols, target_col, district_col,
                    lat_col, lon_col, n_clusters=12, cluster_on="latlong",
                    embedding_cols=None, buffer_km=10, model=None,
                    random_state=42):
    df = df.dropna(subset=feature_cols + [target_col]).reset_index(drop=True)
    df, block_map = build_blocks(df, district_col, lat_col, lon_col,
                                  n_clusters, cluster_on, embedding_cols,
                                  random_state)

    groups = df["block"].values
    n_groups = len(np.unique(groups))
    gkf = GroupKFold(n_splits=n_groups)  # one fold per block

    if model is None:
        model = Ridge(alpha=1.0)

    fold_results = []
    y_true_all, y_pred_all = [], []

    for fold_i, (train_idx, test_idx) in enumerate(gkf.split(df, groups=groups)):
        X_train, X_test = df.loc[train_idx, feature_cols], df.loc[test_idx, feature_cols]
        y_train, y_test = df.loc[train_idx, target_col], df.loc[test_idx, target_col]

        if X_test.empty or X_train.empty:
            continue

        scaler = StandardScaler().fit(X_train)
        X_train_s = scaler.transform(X_train)
        X_test_s = scaler.transform(X_test)

        model.fit(X_train_s, y_train)
        y_pred = model.predict(X_test_s)

        min_dist = min_train_test_distance_km(df, train_idx, test_idx, lat_col, lon_col)
        leakage_flag = min_dist < buffer_km

        fold_r2 = r2_score(y_test, y_pred) if len(y_test) > 1 else np.nan
        fold_results.append({
            "fold": fold_i,
            "block_id": int(df.loc[test_idx, "block"].iloc[0]),
            "n_test_districts": df.loc[test_idx, district_col].nunique(),
            "min_train_test_km": round(min_dist, 1),
            "leakage_flag": leakage_flag,
            "fold_r2": fold_r2,
        })
        y_true_all.extend(y_test.tolist())
        y_pred_all.extend(y_pred.tolist())

    overall_r2 = r2_score(y_true_all, y_pred_all)
    return {
        "overall_blocked_r2": overall_r2,
        "n_blocks": n_groups,
        "fold_details": pd.DataFrame(fold_results),
    }
```

## 6. Naive per-district LODO — for direct side-by-side comparison

```
def run_naive_lodo(df, feature_cols, target_col, district_col, model=None):
    df = df.dropna(subset=feature_cols + [target_col]).reset_index(drop=True)
    groups = df[district_col].values
    n_groups = df[district_col].nunique()
    gkf = GroupKFold(n_splits=n_groups)

    if model is None:
        model = Ridge(alpha=1.0)

    y_true_all, y_pred_all = [], []
    for train_idx, test_idx in gkf.split(df, groups=groups):
        X_train, X_test = df.loc[train_idx, feature_cols], df.loc[test_idx, feature_cols]
        y_train, y_test = df.loc[train_idx, target_col], df.loc[test_idx, target_col]

        scaler = StandardScaler().fit(X_train)
        model.fit(scaler.transform(X_train), y_train)
        y_pred = model.predict(scaler.transform(X_test))

        y_true_all.extend(y_test.tolist())
        y_pred_all.extend(y_pred.tolist())

    return r2_score(y_true_all, y_pred_all)
```

## 7. Run it

```
model = Ridge(alpha=1.0) if MODEL_NAME == "ridge" else RandomForestRegressor(
    n_estimators=300, min_samples_leaf=2, n_jobs=-1)

naive_r2 = run_naive_lodo(df, feature_cols, TARGET_COL, DISTRICT_COL, model)

blocked = run_blocked_cv(
    df, feature_cols, TARGET_COL, DISTRICT_COL,
    LAT_COL, LON_COL,
    n_clusters=N_CLUSTERS, cluster_on=CLUSTER_ON,
    embedding_cols=embedding_cols, buffer_km=BUFFER_KM, model=model,
)

print(f"Crop: {CROP}  |  Model: {MODEL_NAME}  |  Cluster on: {CLUSTER_ON}")
print(f"Naive per-district LODO R^2:      {naive_r2:.4f}")
print(f"Blocked ({blocked['n_blocks']} blocks) LODO R^2:   {blocked['overall_blocked_r2']:.4f}")
print(f"Leakage-inflation gap:            {naive_r2 - blocked['overall_blocked_r2']:.4f}")
```

## 8. Per-fold detail

`leakage_flag=True` means a train district was closer than `BUFFER_KM` to a test district — if many folds are flagged, increase `N_CLUSTERS` (smaller blocks) is wrong direction; actually **decrease** `N_CLUSTERS` (bigger blocks) or raise `BUFFER_KM` tolerance.
