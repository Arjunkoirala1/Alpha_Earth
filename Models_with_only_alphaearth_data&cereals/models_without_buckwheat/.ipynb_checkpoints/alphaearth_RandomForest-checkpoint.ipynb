# RandomForest — alphaearth dataset

Trains **RandomForest** on `alphaearth_cereal.csv` (AlphaEarth embeddings only) to predict cereal yield (Barley, Buckwheat, Maize, Millet, Wheat) using Leave-One-Year-Out and Leave-One-District-Out cross-validation.

Standalone notebook — no dependency on any other model/dataset notebook.

## 1. Imports

```python
import pandas as pd
import numpy as np
from scipy.stats import pearsonr
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import LeaveOneGroupOut, GroupKFold
from sklearn.impute import KNNImputer
from sklearn.metrics import (
    r2_score, mean_squared_error, mean_absolute_error,
    mean_absolute_percentage_error, explained_variance_score, max_error,
)

pd.set_option("display.width", 140)
pd.set_option("display.max_columns", 20)
```

## 2. Config

```python
DATA_PATH = r"C:\Users\SRDC\Desktop\Alpha_earth\Models_separate\alphaearth_cereal.csv"  # put this notebook in the same folder as the CSV, or edit this path
OUT_PATH = "alphaearth_RandomForest_results.csv"
CROPS = ["Barley", "Buckwheat", "Maize", "Millet", "Wheat"]
RANDOM_STATE = 42
MAX_LODO_FOLDS = 15  # GroupKFold size when a crop has more districts than this
```

## 3. Load & clean data

```python
df = pd.read_csv(DATA_PATH)
df.columns = [c.strip() for c in df.columns]
for c in df.columns:
    if df[c].dtype == object:
        df[c] = df[c].str.strip()

target_cols_all = [f"{crop}_{metric}" for crop in CROPS
                    for metric in ["Area_Ha", "Production_Mt", "Yield_MtHa"]]
for c in target_cols_all:
    df[c] = pd.to_numeric(df[c], errors="coerce")

print("Shape:", df.shape)
df.head()
```

## 3b. Handle missing target values

Within each district, interpolate missing yield years linearly (needs ≥1 real year present). Districts with **zero** real data across all years for a crop are left as NaN so they are automatically excluded from training/prediction for that crop — never interpolated, never scored.

A parallel `is_interpolated` flag is kept per crop so that later, during cross-validation, held-out test points are scored **only if their target is a real observed value** — interpolated points are used to help training but never to grade the model. Scoring against an interpolated value would leak information (its "truth" is a linear function of neighboring years already in the training fold), which would silently inflate R².

```python
df = df.sort_values(["District_Key", "Year_AD"]).reset_index(drop=True)

excluded_districts = {}   # crop -> list of districts with zero real data (never fed to the model)
interp_flags = {}         # crop -> boolean array, True = this row's target was filled by interpolation

for crop in CROPS:
    col = f"{crop}_Yield_MtHa"
    was_missing = df[col].isna()

    district_valid_counts = df.groupby("District_Key")[col].transform(lambda s: s.notna().sum())
    zero_mask = district_valid_counts == 0
    excluded_districts[crop] = sorted(df.loc[zero_mask, "District_Key"].unique().tolist())

    def interp_group(s):
        if s.notna().sum() == 0:
            return s  # zero-data district: leave as NaN, will be excluded by the notna() mask later
        return s.interpolate(method="linear", limit_direction="both")

    df[col] = df.groupby("District_Key")[col].transform(interp_group)
    interp_flags[crop] = (was_missing & df[col].notna()).values

    print(f"{crop:10s} | interpolated {interp_flags[crop].sum():3d} row(s) | "
          f"excluded {zero_mask.sum():3d} row(s) from {len(excluded_districts[crop])} "
          f"zero-data district(s): {excluded_districts[crop]}")
```

## 3c. Handle missing feature values (similarity-based, since we have no lat/lon)

`KNNImputer` finds the 5 most *similar* rows (by all other feature values) and fills any gap from them, weighted by similarity. Categorical soil/landform codes (if present in this dataset) are filled with the most common value as a simple fallback.

```python
embed_cols = [c for c in df.columns if c.startswith("A") and c[1:].isdigit()]
numeric_feat_cols = embed_cols
categorical_feat_cols = []

n_missing_before = df[numeric_feat_cols].isna().sum().sum()
if n_missing_before > 0:
    knn_imputer = KNNImputer(n_neighbors=5, weights="distance")
    df[numeric_feat_cols] = knn_imputer.fit_transform(df[numeric_feat_cols])
print(f"Feature imputation: filled {n_missing_before} missing numeric feature value(s) "
      f"via 5-nearest-neighbor similarity")

for c in categorical_feat_cols:
    n_missing_cat = df[c].isna().sum()
    if n_missing_cat > 0:
        df[c] = df[c].fillna(df[c].mode().iloc[0])
        print(f"  {c}: filled {n_missing_cat} missing value(s) with mode")
```

## 4. Feature set

AlphaEarth 64-dim embeddings (`A00`-`A63`) only — no soil chemistry or terrain columns exist in this dataset.

```python
X_all = pd.get_dummies(df[numeric_feat_cols + categorical_feat_cols], columns=categorical_feat_cols)
print("Feature matrix shape:", X_all.shape)
print(f"{len(embed_cols)} embedding dims + {len(numeric_feat_cols) - len(embed_cols)} numeric + "
      f"{X_all.shape[1] - len(numeric_feat_cols)} one-hot soil/landform columns")
```

## 5. Model definition — RandomForest

```python
def make_model():
    return RandomForestRegressor(n_estimators=150, max_depth=8, random_state=RANDOM_STATE, n_jobs=-1)

CV_SCHEMES = {
    "Leave-One-Year-Out": "Year_BS",
    "Leave-One-District-Out": "District_Key",
}
```

## 6. Cross-validation + metrics helpers

```python
def run_cv(y, X, groups, is_interp):
    """Group-aware CV -- true leave-one-out if few groups, else GroupKFold.
    No group (year or district) ever appears in both train and test within a fold.
    Interpolated rows can be used for TRAINING, but are excluded from scoring
    to avoid grading the model against values that were fabricated from
    neighboring years still present in the training fold."""
    n_groups = groups.nunique()
    splitter = LeaveOneGroupOut() if n_groups <= MAX_LODO_FOLDS else GroupKFold(n_splits=MAX_LODO_FOLDS)
    y_true_all, y_pred_all = [], []
    for train_idx, test_idx in splitter.split(X, y, groups):
        model = make_model()
        model.fit(X.iloc[train_idx], y.iloc[train_idx])
        preds = model.predict(X.iloc[test_idx])
        real_mask = ~is_interp[test_idx]  # only score against real observed yields
        y_true_all.extend(y.iloc[test_idx][real_mask].tolist())
        y_pred_all.extend(preds[real_mask].tolist())
    return np.array(y_true_all), np.array(y_pred_all)


def compute_metrics(y_true, y_pred):
    r, _ = pearsonr(y_true, y_pred)
    return {
        "R2": round(r2_score(y_true, y_pred), 3),
        "RMSE": round(mean_squared_error(y_true, y_pred) ** 0.5, 3),
        "MAE": round(mean_absolute_error(y_true, y_pred), 3),
        "MAPE_pct": round(mean_absolute_percentage_error(y_true, y_pred) * 100, 2),
        "Explained_Variance": round(explained_variance_score(y_true, y_pred), 3),
        "Max_Error": round(max_error(y_true, y_pred), 3),
        "Pearson_r": round(r, 3),
    }
```

## 7. Run for every crop x CV scheme

```python
results = []
for crop in CROPS:
    target_col = f"{crop}_Yield_MtHa"
    mask = df[target_col].notna()
    y = df.loc[mask, target_col].reset_index(drop=True)
    X = X_all.loc[mask].reset_index(drop=True)
    groups_df = df.loc[mask].reset_index(drop=True)
    is_interp = interp_flags[crop][mask.values]
    n = len(y)
    n_real = (~is_interp).sum()

    for scheme_name, group_col in CV_SCHEMES.items():
        groups = groups_df[group_col]
        n_groups = groups.nunique()
        if n_groups < 2:
            continue
        y_true, y_pred = run_cv(y, X, groups, is_interp)
        metrics = compute_metrics(y_true, y_pred)
        row = {"Model": "RandomForest", "Crop": crop, "CV_Scheme": scheme_name,
               "N_samples": n, "N_real_samples": n_real, "N_groups": n_groups,
               "N_folds_used": min(n_groups, MAX_LODO_FOLDS)}
        row.update(metrics)
        results.append(row)
        print(f"{crop:10s} | {scheme_name:23s} | R2={metrics['R2']:6.3f} | "
              f"RMSE={metrics['RMSE']:6.3f} | MAE={metrics['MAE']:6.3f} | "
              f"MAPE={metrics['MAPE_pct']:6.2f}%")

results_df = pd.DataFrame(results).sort_values(["Crop", "CV_Scheme"]).reset_index(drop=True)
print("\nDone:", len(results_df), "runs")
```

## 9. Save results

```python
results_df.to_csv(OUT_PATH, index=False)
print(f"Saved to {OUT_PATH}")
```

## 10. Quick visualization — R² by crop, per CV scheme

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 5))
pivot = results_df.pivot(index="Crop", columns="CV_Scheme", values="R2").loc[CROPS]
pivot.plot(kind="bar", ax=ax)
ax.axhline(0, color="black", linewidth=0.8)
ax.set_ylabel("R2")
ax.set_title("RandomForest (alphaearth): Out-of-fold R2 by crop and CV scheme")
plt.tight_layout()
plt.show()
```
